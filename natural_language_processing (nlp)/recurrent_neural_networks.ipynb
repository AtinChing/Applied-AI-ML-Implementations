## Sequences
- RNNs specialize on data that can be represented as sequences, which can just be thought of as groups of tokens that collectively represent data where position is considered important. Think sentences, bunch of numbers over time for something, frames/pictures that make up an entire video.
- This makes them stronger than vanilla ANNs and CNNs, because it allows working with variable size inputs and variable sized outputs, compared to fixed size inputs and outputs from ANNs and CNNs.
- However, they are also extremely effective if the data is fixed-size instead of a sequence.

## How do RNNs work?
- They have input and output vectors just like standard ANNs and CNNs, however every output vector is determined using the input vector **as well as** the history of the past vectors you have inputted, allowing the model to base answers off of entire sequences.
