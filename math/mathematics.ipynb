# Mathematics
- In standard SWE, it is much easier to get away with having a "meh" level understanding of mathematics, to be a good software engineer.
- The same cannot be said for AI/ML engineering, or more specifically, being a good AI/ML engineer. As one of my friends always says, anyone can just import a library and write a couple lines of code to train some sort of model that achieves somewhat decent performance, with a dataset already cleaned and polished perfectly, ready for usage out of the box.
- The reality is that, in the real world, AI/ML engineering is not this simple. To be a good AI/ML engineer, you need to have a strong, low-level, intuitive understanding of the models you are creating and the data you are working with. To really understand these at their absolute core, it is required of you to have a strong understanding and hold on mathematics. You need to have good mathematics.
- This section of the repo purely shows what math concepts you need to know to get ahead in this "course".
- I may expand this section to include mathematical concept explanations of all these topics, at some point in the future.
