# Mathematics
- In standard SWE, it is much easier to get away with having a "meh" level understanding of mathematics, to be a good software engineer.
- The same cannot be said for AI/ML engineering, or more specifically, being a good AI/ML engineer. As one of my friends always says, anyone can just import a library and write a couple lines of code to train some sort of model that achieves somewhat decent performance, with a dataset already cleaned and polished perfectly, ready for usage out of the box.
- The reality is that, in the real world, AI/ML engineering is not this simple. To be a good AI/ML engineer, you need to have a strong, low-level, intuitive understanding of the models you are creating and the data you are working with. To really understand these at their absolute core, it is required of you to have a strong understanding and hold on mathematics. You need to have good mathematics.
- This section of the repo purely shows what math concepts you need to know to get ahead in this "course".
- I may expand this section to include mathematical concept explanations of all these topics, at some point in the future.

## Essential Mathematical Concepts for AI/ML Engineering

### **Linear Algebra** (Critical Foundation)
- **Vectors**: Addition, subtraction, dot product, cross product, norms
- **Matrices**: Multiplication, transpose, inverse, determinant, rank
- **Eigenvalues and Eigenvectors**: Eigendecomposition, diagonalization
- **Singular Value Decomposition (SVD)**: Matrix factorization, dimensionality reduction
- **Vector Spaces**: Linear independence, basis, span, subspaces
- **Matrix Calculus**: Gradients of matrices, chain rule for matrices
- **Special Matrices**: Identity, orthogonal, symmetric, positive definite

### **Calculus** (Optimization Foundation)
- **Derivatives**: Partial derivatives, gradients, directional derivatives
- **Chain Rule**: Composite function differentiation (backbone of backpropagation)
- **Multivariable Calculus**: Gradients, Hessians, Jacobians
- **Optimization**: Critical points, maxima/minima, saddle points
- **Vector Calculus**: Divergence, curl, gradient fields
- **Taylor Series**: Function approximation, linearization

### **Probability and Statistics** (Data Understanding)
- **Probability Fundamentals**: Sample spaces, events, conditional probability
- **Bayes' Theorem**: Prior, posterior, likelihood
- **Random Variables**: Discrete, continuous, expectation, variance
- **Probability Distributions**: 
  - Discrete: Bernoulli, binomial, multinomial, Poisson
  - Continuous: Gaussian/normal, exponential, beta, gamma
- **Central Limit Theorem**: Sampling distributions, normal approximation
- **Statistical Inference**: Confidence intervals, hypothesis testing
- **Maximum Likelihood Estimation (MLE)**: Parameter estimation
- **Bayesian Statistics**: Prior/posterior distributions, Bayesian inference

### **Optimization Theory** (Training Foundation)
- **Convex Optimization**: Convex sets, convex functions, global minima
- **Gradient Descent**: Steepest descent, convergence analysis
- **Constrained Optimization**: Lagrange multipliers, KKT conditions
- **Non-convex Optimization**: Local minima, saddle points, escaping strategies
- **Stochastic Optimization**: SGD, momentum, adaptive methods
- **Duality Theory**: Primal-dual problems, strong/weak duality

### **Information Theory** (Learning Theory)
- **Entropy**: Shannon entropy, conditional entropy, joint entropy
- **Mutual Information**: Information gain, feature selection
- **Cross-entropy**: Classification loss functions
- **Kullback-Leibler (KL) Divergence**: Distribution comparison, regularization
- **Information Gain**: Decision trees, feature importance

### **Graph Theory** (Network Architectures)
- **Graph Basics**: Vertices, edges, adjacency matrices
- **Graph Types**: Directed, undirected, weighted, bipartite
- **Graph Traversal**: BFS, DFS, shortest paths
- **Spectral Graph Theory**: Graph Laplacian, eigenvalues of graphs
- **Network Analysis**: Centrality measures, clustering coefficients

### **Numerical Methods** (Computational Implementation)
- **Numerical Linear Algebra**: LU decomposition, QR decomposition, iterative methods
- **Numerical Integration**: Quadrature rules, Monte Carlo integration
- **Numerical Optimization**: Newton's method, quasi-Newton methods
- **Finite Differences**: Numerical differentiation, gradient checking
- **Interpolation**: Polynomial, spline, radial basis functions
- **Root Finding**: Newton-Raphson, bisection method

### **Advanced Topics** (Deep Learning Specifics)
- **Functional Analysis**: Function spaces, norms, inner products
- **Differential Geometry**: Manifolds, tangent spaces, Riemannian geometry
- **Fourier Analysis**: Fourier transforms, frequency domain analysis
- **Measure Theory**: Integration theory, probability measures
- **Topology**: Open/closed sets, continuity, compactness
- **Complex Analysis**: Complex derivatives, contour integration

### **ML-Specific Mathematical Concepts**
- **PAC Learning**: Probably Approximately Correct learning theory
- **VC Dimension**: Vapnik-Chervonenkis dimension, generalization bounds
- **Rademacher Complexity**: Generalization theory, uniform convergence
- **Kernel Methods**: Reproducing kernel Hilbert spaces, kernel trick
- **Regularization Theory**: Bias-variance tradeoff, Tikhonov regularization
- **Concentration Inequalities**: Hoeffding, Chernoff bounds

### **Deep Learning Mathematics**
- **Automatic Differentiation**: Forward mode, reverse mode (backpropagation)
- **Tensor Operations**: Multi-dimensional arrays, Einstein notation
- **Activation Function Analysis**: Lipschitz continuity, universal approximation
- **Loss Function Properties**: Convexity, smoothness, convergence guarantees
- **Normalization Techniques**: Batch norm, layer norm mathematical foundations
- **Attention Mechanisms**: Scaled dot-product attention, multi-head attention math
