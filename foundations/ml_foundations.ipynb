### Curve fitting (function fitting)
- This is the process of defining or determining a function that best approximates the relationship between outputs/dependent variables, and inputs/independent variables, based on a set of data points. 
  - In 2D linear regression, this is just the y = f(x) function.
  - In neural networks, the function that is represented by the neural network that maps inputs to outputs, is the function on its own. And so on.
- Essentially, you are trying to find a curve/mathematical function that best "fits"/approximates a series of data points (inputs and outputs), and minimizes the overall error in its outputs.
- This is obviously done by training our model on a specific set of training data, before experimenting on test data, before serving it to the real world on completely unseen data.
  - The goal is to have a generalized model that performs well on unseen data.
- Sometimes, you might end up with functions that fit TOO well or fit TOO loosely to the training data. This can cause issues when it encounters unseen data...
  - **Underfitting**: The function fits loosely on training data, and performs badly on it, and therefore performs badly on unseen data too.
    - We say that the model has failed to capture the complex underlying patterns in the data, as the model is too simple for the problem,, just not sufficiently complex.
  - **Overfitting**: The function fits tightly on training data and performs exceptionally on it, to the point where it practically memorizes the training data, and therefore does not generalize well on unseen data.
- To understand this better, we need to look at statistical terms: bias and variance.

#### Bias
- This is the error between the average predictions made by the model (i.e. E[g(x)]) and the ground truth function f(x)
- Bias reflects how well your model class can fit the true function at all. (Model class = all functions that your type of model can have, e.g. linear regression would be all straight lines, planes, hyperplanes, no curves).
- To compute:
  - $\mathrm{Bias}^2 = \mathbb{E}\left[ (\mathbb{E}[g(x)] - f(x))^2 \right]$ 
    - where $\mathbb{E}[g(x)]$ is the average prediction.
- High bias indicates underfitting. The model is too basic, can't approximate the data well.

#### Variance
- Variance measures how much the model's predictions fluctuate for different training sets.
- It quantifies the expected squared deviation of a single model's prediction from the average prediction, given a specific input.
- Indicates how sensitive the model is to the noise or variations in the training data.
- To compute:  
  - $\mathrm{Variance} = \mathbb{E}\left[ \left( g(x) - \mathbb{E}[g(x)] \right)^2 \right]$
    - where $g(x)$ is a prediction from a single model trained on a possible dataset (of the same type) and $\mathbb{E}[g(x)]$ is the average prediction over all such models.
  - $\mathbb{E}[g(x)]$: average prediction

#### Detecting Underfitting or Overfitting
- We can try to visualize the fit of the model on the test data (how predicted values map vs actual values), usually via graphs.
- We can study the bias-variance trade off:
  - If a model has high bias and low variance, the model underfits the data.
  - And vice-versa, the model would overfit the data.
- When both bias and variance are low, we have a good approximation of the true function!
- The goal is to always try to minimize the **error** of the model.
- The decomposition of the model's error can be written as:
  - $\text{Error of the model} = \mathrm{Bias}^2 + \mathrm{Variance} + \text{Irreducible Error}$

#### K-fold Cross Validation
- This is a technique used to compare models and prevent overfitting in models.
- A resampling procedure to help the model to generalize well
- Has a single parameter called k for the number of partitions
- Procedure for k-fold cross validation:
  1. Randomize the dataset and create k equal size partitions
  2. Use k-1 partitions for training the model
  3. Use the k-th partition for testing and evaluating the model
  4. Record the evaluation scores (such as MSE, R^2) in each iteration
  5. Iterate k times with a different subset reserved for testing purpose each time.
  6. Select the model with the highest performance score and average the evaluation scores.
- Some commonly used variations on cross-validation are stratified k-fold, leave-one-out, and repeated k-fold are available in scikit-learn.
- Highest performance score means the lowest error!
- You can also compute the Bias and Variance for each fold to get a complete picture about the model.
- You can use Mean Squared Error (MSE) and R^2 evaluation scores to evaluate the performance of a model during each iteration.
