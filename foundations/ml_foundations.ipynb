### Curve fitting (function fitting)
- This is the process of defining or determining a function that best approximates the relationship between outputs/dependent variables, and inputs/independent variables, based on a set of data points. 
  - In 2D linear regression, this is just the y = f(x) function.
  - In neural networks, the function that is represented by the neural network that maps inputs to outputs, is the function on its own. And so on.
- Essentially, you are trying to find a curve/mathematical function that best "fits"/approximates a series of data points (inputs and outputs), and minimizes the overall error in its outputs.
- This is obviously done by training our model on a specific set of training data, before experimenting on test data, before serving it to the real world on completely unseen data.
  - The goal is to have a generalized model that performs well on unseen data.
- Sometimes, you might end up with functions that fit TOO well or fit TOO loosely to the training data. This can cause issues when it encounters unseen data...
  - **Underfitting**: The function fits loosely on training data, and performs badly on it, and therefore performs badly on unseen data too.
  - **Overfitting**: The function fits tightly on training data and performs exceptionally on it, to the point where it practically memorizes the training data, and therefore does not generalize well on unseen data.
