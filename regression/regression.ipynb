# Regression models
- Please reference ml_foundations.ipynb before proceeding with this section.
- A regression model is any model that attempts to find the relationship between a continuous-valued dependent variable (outputs) and one or more indpendent variables (inputs).

## Methods to measure performance
- There are many ways to measure performance of regression models. This can be:
  - Mean Squared Error (MSE)
  - $R**2$ (Coeffcient of Determination)
- Bias and variance tell us the cause of error, but don't directly measure model performance. 
- MSE and $R**2$ are more practical measures that give a direct assessment of how well the mdoel fits the data. They're also easier to compute!
