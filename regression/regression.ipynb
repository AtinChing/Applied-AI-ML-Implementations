# Regression models
- Please reference ml_foundations.ipynb before proceeding with this section.
- A regression model is any model that attempts to find the relationship between a continuous-valued dependent variable (outputs) and one or more indpendent variables (inputs).

## Methods to measure performance
- There are many ways to measure performance of regression models. This can be:
  - Mean Squared Error (MSE)
  - $R^2$ (Coeffcient of Determination)
- Bias and variance tell us the cause of error, but don't directly measure model performance. 
- Mean Squared Error ($\mathrm{MSE}$) and $R^2$ are practical measures that give a direct assessment of how well the model fits the data, and are straightforward to compute.

 ## Mean Squared Error ($\mathrm{MSE}$)
 
 - Assume a dataset $D$ with $n$ observations.
 - For the $i^{th}$ observation, let $x_i$ denote the input, $y_i = f(x_i)$ be the true (target) value, and $\hat{y}_i = g(x_i)$ be the predicted value from the model.
 - $\mathrm{MSE}$ is calculated as:
 
 $$ \mathrm{MSE} = \frac{1}{n} \sum_{i=1}^n (y_i - \hat{y}_i)^2 $$
 
 - The $\mathrm{MSE}$ can be used to evaluate how well the model fits both the training data and the test data.

## Coefficient of Determination ($R^2$)

- $R^2$ represents the strength of the relationship between independent variables (input) and the dependent variable (output).
- A higher value of $R^2$ indicates a stronger relationship between the independent variables and the dependent variable.
- A normalized measure of variance in the output with respect to the input.
- Easier to interpret because its value does not depend on the scale of the data.
- Serves as an indicator of the goodness of fit for regression models.
- Range: $0.0 \leq R^2 \leq 1.0$

*(Note: $R^2$ can be negative when a model performs worse than predicting the mean; for effective models it typically falls in this range.)*

**Formulas:**

$$R^2 = 1 - \frac{\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}{\sum_{i=1}^{n}(y_i - \bar{y})^2}$$

Equivalently, in terms of MSE:
$$R^2 = 1 - \frac{n \times \mathrm{MSE}}{\sum_{i=1}^{n}(y_i - \bar{y})^2}$$

Where:
- $n$ = total number of observations
- $y_i$ = actual observed value for the $i$-th observation
- $\hat{y}_i$ = predicted value from the model for the $i$-th observation
- $\bar{y}$ = mean of the actual observed values
- $\mathrm{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$


## What's better?
- $R^2$ can increase as more predictors are added to a model, even if they don't really improve the model by being added. This gives an illusion of improvement.
- $MSE$ doesn't necessarily increase with more predictors, making it a more truthful metric to use when playing around with model complexity. So therefore, if MSE decreases as predictors are added, it typically indivates genuine improvement (but overfitting should still be something to look out for as its always a concern).
