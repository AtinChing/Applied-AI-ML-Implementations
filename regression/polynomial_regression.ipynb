# Polynomial Regression
Polynomial regression models the relationship between the dependent variable $y$ and the independent variable $x$ as an $n$th degree polynomial:

$$
f(x) = \omega_0 + \omega_1 x + \omega_2 x^2 + \omega_3 x^3 + \cdots + \omega_n x^n + \epsilon
$$

Here:
- $y$ (dependent values) is predicted by the function $f(x)$.
- $\omega_0$ is the intercept.
- $\omega_1, \omega_2, ..., \omega_n$ are the model parameters (weights) for each corresponding power of $x$.
- $\epsilon$ denotes the error term.

Alternatively, by relabeling variables, it can be written as:

$$
f(z) = \omega_0 + \omega_1 z_1 + \omega_2 z_2 + \omega_3 z_3 + \cdots + \omega_n z_n + \epsilon
$$

With respect to the weights, polynomial regression remains in the same family of linear models.

### Measure loss
- The best function to use for polynomial regression is Root Mean Square error. 
$$
E_{\mathrm{RMS}} = \sqrt{\frac{2 E(\mathbf{w}^\star)}{N}}
$$

- where:
  - $E(\mathbf{w}^\star)$ is the sum-of-squares error evaluated at the optimal weights $\mathbf{w}^\star$,
  - $N$ is the number of data points.
- RMSE tells us how concentrated the data is around the line of best fit.

### Overfitting
- You need to prevent overfitting, and there are multiple ways to do this with polynomial regression:
  - Cross validation
  - Increase sample size for training data
  - Remove features
  - Early stopping
  - Regularization 

### Regularization
- This is a very interesting technique.
- Technique to prevent overfitting my adding a penalty term to the loss function, that actively discourages overly complex models.
- 2 types:
  - L1 regularization (Lasso)
  - L2 regularization (Ridge).

#### Lasso Regression
Lasso Regression

- $M$ : number of samples in $D$
- $N$ : number of attributes
- $c$ is the maximum allowable value for the L1 norm of the coefficients.
  - It can be small for more aggressive regularization, leading to a simpler model.
  - It can be large for less regularization, allowing more features to have non-zero coefficients.

In Lasso regression, the loss function is revised by adding a penalty.

Penalty = sum of the absolute of the magnitude of the coefficients, multiplied by the penalty term $(\lambda)$:
$$
\lambda \sum_{j=0}^{N} |w_j|
$$

The constraint $c$ helps in regularizing the model by shrinking some weights to exactly 0.

For some $c > 0$,
$$
\sum_{j=0}^{N} |w_j| < c
$$
Constraint on regression coefficients.

#### Ridge Regression

- M : number of samples in D  
- N : number of attributes

In ridge regression, large weights (coefficients) are penalized more.  
The loss function for ridge regression is revised by adding a penalty term:
$$
\sum_{i=1}^{M} (y_i - \hat{y}_i)^2 = \sum_{i=1}^{M} \bigg( y_i - \sum_{j=0}^{N} w_j \ast x_{ij} \bigg)^2
$$

Penalty = sum of the square of the magnitude of the coefficients, multiplied by the penalty term $(\lambda)$:
$$
\lambda \sum_{j=0}^{N} w_j^2
$$

So the total cost function becomes:
$$
\sum_{i=1}^{M} (y_i - \hat{y}_i)^2 + \lambda \sum_{j=0}^{N} w_j^2
$$

For some $c > 0$,
$$
\sum_{j=0}^N w_j^2 < c
$$
This above is a constraint on regression coefficients.

The quadratic penalty grows faster for larger weights, which disproportionately affects larger weights compared to smaller ones.
