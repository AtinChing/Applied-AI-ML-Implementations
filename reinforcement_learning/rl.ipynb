# What is Reinforcement Learning?
- Separate paradigm of machine learning from unsupervised and supervised learning etc.
- Operates on a reward-penalty system and learns in the form of being an agent that is interacting with an environment/world, and learning what does and doesn't work.

## Learning the components
- There are multiple components of RL that we need to learn for RL to work well.

#### Agent
- An AI agent is an autonomous entity capable of perceiving its environment and taking actions independently.
- Key characteristics of an agent (on a theoretical level) include:
  - Autonomy: operates on its own, making its own decisions.
  - Goals: has objectives or tasks it tries to accomplish.
  - Perception: can sense or observe its surroundings or receive inputs in some way.
  - Action: can affect its environment through its choices by taking certain actions (e.g. tool calls)
  - Adaptation: can modify its behavior based on experience or changes in the environment.
- Agents can be found in many domains, ranging from robotics and software systems to everyday life.

#### Environment
- The world/context/model that the agent interacts with.

#### State
- A complete description of the state of the world or environment. Every piece of info about the world is included in this, but it may not be entirely visible to the agent in all scenarios.

#### Observation
- A partial description of the state, and can omit information about the environment. This is what the agent perceives of the world.
- This leads to 2 types of observations:
  - Fully-observed environments: When an agent is fully able to observe and perceive the ENTIRE state of an environment. (E.g. Chess, where you can see all pieces i.e. the entire state).
  - Partially-observed environments: When an agent can only partially observe the entire state of an environemnt. (E.g. Poker, where other players cards are hidden).

#### Reward (function)
- The score/outcome that the agent achieves by taking certain actions. Positive reward is meant to reinforce certain behaviour (whatever actions led to that outcom/state), and negative reward (punishment) is meant to inhibit certain behaviour.
- The reward function is an abstract definition of good behavior. It can be used to define whether a certain state or state-action pair is good or bad in relation to the goal of the agent.

#### Policy
- Represented by 𝜋.
- It is the behaviour function of the agent. It literally defines how an agent is going to respond given its current state and its possible next actions.
- It maps states/observations to actions. In other words, given what the agent knows/perceives, the policy tells it what action to take next.
- This policy π can be either:
  - Deterministic (selects a specific action for each state, so it learns hard-coded actions for each state, and the same state always means/outputs the same action). 
  $$\pi : S \rightarrow A$$
  - Stochastic (gives a probability distribution over actions for each state).
  $$\pi : S \rightarrow P(A)$$
- Policies can be represented in many ways: as tables (for small environments), as functions (learned with neural networks in deep RL), or as sets of rules.

#### Value/utility function
- How good is each state or action.
- Gives a mapping from states to expected rewards.
$$U : S \rightarrow \mathbb{R}$$

## Markov Decision Process (MDP)
- Standard mathematical framework used to model RL systems and problems. 
- An MDP is a 5-tuple, $\langle S, A, R, P, \rho_0 \rangle$

**Components:**
- $S$ - set of valid states
- $A$ - set of valid actions
- $R: S \times A \times S \rightarrow \mathbb{R}$ - reward function
- $P: S \times A \rightarrow \mathcal{P}(S)$ - transition probability
- $\rho_0: S \rightarrow \mathbb{R}$ - initial state distribution

**Probability Notations:**
- $P(s' | s, a)$ is the probability of transitioning into state $s'$ if you start in state $s$ and take action $a$
- $\rho_0(s)$ is the probability of starting in state $s$

#### Action space
- Set of valid actions in a given environment.
- There are 2 types:
  - Discrete: only a finite number of moves are available to the agent. Example: Tic-Tac-Toe, Chess, Go, etc.
  - Continuous: In continuous spaces, actions are real-valued vectors. Example: continuous control tasks, like driving a car or controlling a robot.

## Where can RL be applied?
In any scenario where: 
  - Collection of data of good or intended behaviour is difficult
  - There is a problem to optimize for in some type of environment or context or world with which an agent can interact.

### Challenges of RL
1. Credit Assignment Problem: how do you know what led to a
reward?
2. Delayed Consequences: actions in the world have long-term
consequences that don’t show up immediately.
3. Reward Sparsity: rewards are sometimes given infrequently.
4. Exploration vs. Exploitation: doing what works vs. trying new
