# Exploratory Data Analysis
- Relates to looking carefully at data and trying to determine any patterns or trends, outliers or unusual points or using viualizations (like charts and tables) to determine just any characteristics. This is very important before moving to other important machine learning techniques.
- EDA helps summarize the dataset’s information, aiming to get a good understanding of the data. It's an important early part of working with data in whatever pipeline (likely ML) you are working in that involves that data.

## The motivation: datasets
- Datasets are arguably the most important factor in building ML models. The reason for this is:
  - Completeness: the level to which a dataset contains all relevant features or observations necessary for learning for a given task. Incomplete datasets may also include missing labels or sparsity in attribute values, and developers should stay cautious of this.
  - Size: to have sufficient number of observations.
  - Validity: dataset that contains accurate labels (there are many unlabeled datasets), clean data (free of outliers and errors), relevant data to the problem. 

## Linearity
- Datasets with linear behaviour can learn linear decision boundaries, whilst nonlinear datasets can learn non linear decision boundaries through non-linear alogrithms such as tree0based algorithms.
- A decision boundary is where the algo draws a classification line to generate an output label or a fitted line to predict a value.
- Understanding linearity can help inform decisions when choosing ML techniques or algorithms

## Outliers
- Samples in a dataset that are abnormally distant from other samples.
- Can harm overall accuracy of the model trained on the data.
- Some methods:
  - Background knowledge (example: age of a person can not be 0 or negative)
  - Visualization (such as scatter plot)
  - Data Analysis (statistical approach, box plot)
  - ML algorithms such as One-Class-SVM, Local outlier factor, Isolation Forest

## Skew in data
- Skewness is a method to measure the asymmetry of a dataset.
  - In a way, shows the direction of outliers/where they lie.
- Standard deviation is a stat that measures the dispersion of data relative to the mean (how much does the data tend to vary from the mean?)
  - Higher S.D. means data crowds around the mean more.
- Median is the value in the middle of the dataset, with 50% of values smaller than it or equal, and 50% larger than it.

### Right skewed vs Left Skewed data
- If median < mean, data is skewed to the right. This means most data points **fall on the left side**, with fewer appearing as you move towards the right. This should be intuitive as the median being to the left of the mean means more data points appear around there, and therefore most of the outliers lie on the right side of the distribution.
  - They are described as having a longer tail on the right side of the distiribution of data.
  - If median > mean, then vice-versa.

### The importance of skewed data
- Many stats models assume that data is normally distributed, and skewness obviously violates this assumption. This can lead to biased parameter estiamtes or inaccurate predictions.
  -  Biased parameter estimations (mean, variance, regression coefficients, etc experience a bias)
- Skewed data can be transformed so we can approximate a more symmetric distribution. These include logarithmic, square root or inverse transformations too.
